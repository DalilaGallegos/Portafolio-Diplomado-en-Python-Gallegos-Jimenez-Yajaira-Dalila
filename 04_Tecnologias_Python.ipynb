A continuación se presenta un ejemplo sencillo relacionado con calificaciones de estudiantes.

```python
estudiantes = [
    {"nombre": "Ana", "calificacion": 9.2},
    {"nombre": "Luis", "calificacion": 7.5},
    {"nombre": "María", "calificacion": 8.7},
    {"nombre": "Carlos", "calificacion": 5.8},
    {"nombre": "Sofía", "calificacion": 9.6}
]

def clasificar_estudiante(calificacion):
    if calificacion >= 6:
        return "Aprobado"
    else:
        return "Reprobado"

for estudiante in estudiantes:
    estado = clasificar_estudiante(
        estudiante["calificacion"]
    )

    print(
        estudiante["nombre"],
        "-",
        estudiante["calificacion"],
        "-",
        estado
    )
```

## 1.4 Explicación del resultado

En este ejemplo se utiliza una **lista de diccionarios** para almacenar el nombre y la calificación de diferentes estudiantes.

También se creó una función llamada `clasificar_estudiante()`, que recibe una calificación y utiliza una estructura condicional para determinar si el estudiante aprobó o reprobó.

Posteriormente, mediante un ciclo `for`, el programa recorre todos los estudiantes y muestra su nombre, calificación y resultado.

El ejemplo demuestra algunos fundamentos de Python:

- Variables.
- Listas.

Estos elementos constituyen la base para desarrollar programas más complejos.

# 4. Scikit-learn

## 4.1 ¿Qué es Scikit-learn?

Scikit-learn es una biblioteca de Python especializada en **Machine Learning**.

En este ejemplo se desarrollará un pequeño problema de clasificación relacionado con el desempeño académico.

## 4.2 Creación de un conjunto de datos de ejemplo

Para demostrar el funcionamiento de Scikit-learn se generará un conjunto de datos sintético de estudiantes.

Cada estudiante tendrá las siguientes variables:

- Horas de estudio.
- Porcentaje de asistencia.
- Porcentaje de tareas entregadas.
- Promedio anterior.

La variable objetivo será:

- `0`: estudiante en riesgo de reprobar.
- `1`: estudiante con resultado de aprobación.

Los datos se utilizarán únicamente como ejemplo educativo para demostrar el funcionamiento de un modelo de clasificación.

```python
import numpy as np

np.random.seed(42)

n_estudiantes = 500

horas_estudio = np.random.uniform(
    0,
    10,
    n_estudiantes
)

asistencia = np.random.uniform(
    50,
    100,
    n_estudiantes
)

tareas = np.random.uniform(
    40,
    100,
    n_estudiantes
)

promedio_anterior = np.random.uniform(
    5,
    10,
    n_estudiantes
)

ruido = np.random.normal(
    0,
    0.8,
    n_estudiantes
)

puntaje = (
    0.30 * horas_estudio
    + 0.04 * asistencia
    + 0.03 * tareas
    + 0.50 * promedio_anterior
    + ruido
)

aprobo = (
    puntaje >= 10.5
).astype(int)

df_ml = pd.DataFrame({
    "Horas_estudio": horas_estudio,
    "Asistencia": asistencia,
    "Tareas": tareas,
    "Promedio_anterior": promedio_anterior,
    "Aprobo": aprobo
})

df_ml.head()
```

```python
print("Cantidad de registros:", len(df_ml))

print("\nDistribución de la variable objetivo:")

print(
    df_ml["Aprobo"].value_counts()
)

print("\nPorcentaje:")

print(
    (
        df_ml["Aprobo"]
        .value_counts(normalize=True)
        * 100
    ).round(2)
)
```

El conjunto sintético contiene 500 registros.

Cada registro representa un estudiante y la variable `Aprobo` contiene el resultado que intentará predecir el modelo.

El objetivo será utilizar las cuatro variables predictoras para clasificar a cada estudiante.

## 4.3 Separación de variables

Las variables predictoras serán almacenadas en `X`, mientras que la variable objetivo será almacenada en `y`.

```python
X = df_ml.drop(
    columns=["Aprobo"]
)

y = df_ml["Aprobo"]
```

## 4.4 División en entrenamiento y prueba

Los datos se dividirán en dos conjuntos:

- 80 % para entrenamiento.
- 20 % para prueba.

El conjunto de entrenamiento será utilizado por el algoritmo para aprender y el conjunto de prueba permitirá evaluar su desempeño con registros que no participaron durante ese proceso.

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Registros de entrenamiento:", len(X_train))
print("Registros de prueba:", len(X_test))
```

## 4.5 Modelo de clasificación

Se utilizará un modelo de **Regresión Logística**.

A pesar de su nombre, la Regresión Logística se utiliza para resolver problemas de clasificación.

El modelo intentará aprender la relación existente entre las variables académicas y el resultado de aprobación.

```python
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression

modelo_ml = Pipeline([
    (
        "escalador",
        StandardScaler()
    ),
    (
        "modelo",
        LogisticRegression(
            max_iter=1000
        )
    )
])

modelo_ml.fit(
    X_train,
    y_train
)

print("Modelo entrenado correctamente.")
```

La utilización de `StandardScaler` permite colocar las variables en escalas comparables.

El Pipeline aplica automáticamente el escalamiento y posteriormente entrena la Regresión Logística.

## 4.6 Predicciones y evaluación

```python
from sklearn.metrics import (
    accuracy_score,
    classification_report,
    confusion_matrix
)

y_pred = modelo_ml.predict(
    X_test
)

accuracy_ml = accuracy_score(
    y_test,
    y_pred
)

print(
    f"Accuracy: {accuracy_ml:.2%}"
)
```

```python
print(
    classification_report(
        y_test,
        y_pred,
        target_names=[
            "En riesgo",
            "Aprueba"
        ]
    )
)
```

```python
matriz_ml = confusion_matrix(
    y_test,
    y_pred
)

print("Matriz de confusión:")
print(matriz_ml)
```

## 4.7 Explicación del resultado

El modelo utiliza patrones presentes en los datos de entrenamiento para clasificar registros que no participaron durante su aprendizaje.

El **Accuracy** representa la proporción total de predicciones correctas.

El reporte de clasificación permite analizar también métricas como Precision, Recall y F1-score.

La matriz de confusión permite identificar cuántos estudiantes fueron clasificados correctamente y qué tipos de errores produjo el modelo.

Este ejemplo demuestra el flujo básico de un proyecto de Machine Learning con Scikit-learn:

**datos → separación → entrenamiento → predicción → evaluación**

# 5. TensorFlow

## 5.1 ¿Qué es TensorFlow?

TensorFlow es una biblioteca utilizada para desarrollar y entrenar modelos de aprendizaje automático y aprendizaje profundo.

Permite construir redes neuronales capaces de trabajar con diferentes tipos de información, incluyendo datos tabulares, imágenes, texto y audio.

TensorFlow incorpora Keras como una interfaz que facilita la construcción de redes neuronales mediante capas.

## 5.2 Aplicaciones en aprendizaje profundo

TensorFlow puede utilizarse en aplicaciones como:

- Clasificación de imágenes.
- Reconocimiento de objetos.
- Procesamiento de lenguaje natural.
- Predicciones basadas en datos.
- Redes neuronales profundas.
- Análisis de secuencias.
- Reconocimiento de voz.

En el Notebook 3 se utilizó TensorFlow para construir una Red Neuronal Convolucional capaz de clasificar imágenes de CIFAR-10.

En este apartado se utilizará un ejemplo más sencillo con datos tabulares para mostrar la construcción básica de una red neuronal.

## 5.3 Preparación de los datos

Se utilizará el mismo conjunto de estudiantes creado anteriormente para comparar el procedimiento con una red neuronal.

Las variables serán estandarizadas antes de ser procesadas por TensorFlow.

```python
from sklearn.preprocessing import StandardScaler

scaler_tf = StandardScaler()

X_train_tf = scaler_tf.fit_transform(
    X_train
)

X_test_tf = scaler_tf.transform(
    X_test
)

print("Datos preparados correctamente.")
```

El escalador se ajustó únicamente utilizando los datos de entrenamiento.

Posteriormente, esos mismos parámetros se utilizaron para transformar los datos de prueba.

Este procedimiento evita utilizar información del conjunto de prueba durante el entrenamiento.

## 5.4 Construcción de la red neuronal

La red tendrá:

- Una capa de entrada con cuatro variables.
- Una primera capa oculta de 16 neuronas.
- Una segunda capa oculta de 8 neuronas.
- Una capa de salida de una neurona.

Las capas ocultas utilizarán la función de activación `ReLU`.

La salida utilizará `sigmoid`, adecuada para clasificación binaria.

```python
import tensorflow as tf

tf.random.set_seed(42)

modelo_tf = tf.keras.Sequential([

    tf.keras.layers.Input(
        shape=(4,)
    ),

    tf.keras.layers.Dense(
        16,
        activation="relu"
    ),

    tf.keras.layers.Dense(
        8,
        activation="relu"
    ),

    tf.keras.layers.Dense(
        1,
        activation="sigmoid"
    )
])

modelo_tf.summary()
```

El resumen del modelo permite observar las capas utilizadas y la cantidad de parámetros que la red neuronal aprenderá durante el entrenamiento.

## 5.5 Compilación

La red utilizará:

- `Adam` como optimizador.
- `binary_crossentropy` como función de pérdida.
- `accuracy` como métrica.

`binary_crossentropy` es adecuada porque el problema contiene únicamente dos posibles categorías.

```python
modelo_tf.compile(
    optimizer="adam",
    loss="binary_crossentropy",
    metrics=["accuracy"]
)
```

## 5.6 Entrenamiento

La red será entrenada durante 30 épocas utilizando el conjunto de entrenamiento.

Se reservará el 20 % de estos datos para validación.

```python
historial_tf = modelo_tf.fit(
    X_train_tf,
    y_train,
    epochs=30,
    batch_size=16,
    validation_split=0.20,
    verbose=0
)

print("Entrenamiento finalizado.")
```

## 5.7 Evolución del entrenamiento

Se visualizará la evolución del Accuracy durante las épocas.

```python
plt.figure(figsize=(8, 5))

plt.plot(
    historial_tf.history["accuracy"],
    label="Entrenamiento"
)

plt.plot(
    historial_tf.history["val_accuracy"],
    label="Validación"
)

plt.title(
    "Accuracy de la red neuronal"
)

plt.xlabel("Época")
plt.ylabel("Accuracy")
plt.legend()

plt.show()
```

La gráfica permite observar cómo cambia el desempeño de la red neuronal durante el proceso de aprendizaje.

Comparar entrenamiento y validación ayuda a comprobar si el modelo mejora de manera similar con datos que no utiliza directamente para actualizar sus parámetros.

## 5.8 Evaluación de la red neuronal

```python
loss_tf, accuracy_tf = modelo_tf.evaluate(
    X_test_tf,
    y_test,
    verbose=0
)

print(
    f"Loss de prueba: {loss_tf:.4f}"
)

print(
    f"Accuracy de prueba: {accuracy_tf:.2%}"
)
```

## 5.9 Predicciones

La salida de la red neuronal corresponde a una probabilidad entre 0 y 1.

Se utilizará un umbral de 0.5 para convertir estas probabilidades en categorías.

```python
probabilidades_tf = modelo_tf.predict(
    X_test_tf,
    verbose=0
).flatten()

predicciones_tf = (
    probabilidades_tf >= 0.5
).astype(int)

resultados_tf = pd.DataFrame({
    "Real": y_test.values[:10],
    "Probabilidad": probabilidades_tf[:10],
    "Predicción": predicciones_tf[:10]
})

resultados_tf
```

## 5.10 Explicación del resultado

La red neuronal genera inicialmente probabilidades.

Una probabilidad mayor o igual a 0.5 se interpreta como clase `1`, mientras que valores inferiores se clasifican como `0`.

El Accuracy obtenido sobre el conjunto de prueba permite conocer qué proporción de registros fue clasificada correctamente.

Este ejemplo muestra cómo TensorFlow permite construir una red neuronal desde sus capas de entrada hasta la evaluación final.

Obtener una predicción o un Accuracy elevado no significa automáticamente que un modelo sea adecuado para cualquier situación. Es necesario analizar los datos, los errores y el contexto en el que se pretende utilizar la solución.

En conjunto, estas tecnologías proporcionan herramientas que pueden utilizarse para transformar datos en información y desarrollar soluciones orientadas a problemas reales.
